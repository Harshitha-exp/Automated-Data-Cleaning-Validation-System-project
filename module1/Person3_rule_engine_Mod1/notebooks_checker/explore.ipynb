```python
# Farmer file test: a fake dataset with planted errors, to check that the 
# rule engine's PII and format rules catch them. Uses no RMA data:
# the crop and the cause names below are typed in by hand
import random 
from faker import Faker

# maked faker repeatable, so evryone gets the same file 
Faker.seed(1)
random.seed(1)
fake = Faker("en_US")

# RMA-style values, so the file reads like insurance claims
crops = ["Corn", "Soybeans", "Wheat", "Cotton"]
causes = ["Drought", "Excess Moisture/Precip/Rain", "Hail", "Freeze"]

# Build 50 fake farmers
rows = []
for i in range(50):
    rows.append({
        "farmer_id": f"F{1000 + i}",
        "full_name": fake.name(),
        "email": fake.email(),
        "phone": fake.phone_number(),
        "address": fake.address().replace("\n", ", "),
        "date_of_birth": fake.date_of_birth(minimum_age=25, maximum_age=80).strftime("%Y-%m-%d"),
        "county_fips": random.choice(["01001", "19153", "17031"]),
        "crop": random.choice(crops),
        "cause_of_loss": random.choice(causes),
        "claim_amount": round(random.uniform(500, 50000), 2),
    })
syn = pd.DataFrame(rows)

# Planted errors: the answer key
syn.loc[7, "date_of_birth"] = "12/05/1980"     # a second date format
# lost leading zero
syn.loc[15, "county_fips"] = "1001"            
# extreme outlier
syn.loc[20, "claim_amount"] = 9_999_999        
# missing value (not one of the engine's rules)
syn.loc[25, "crop"] = None                     

# Save the fake file (optional): lands in profiling_rules/data/
syn.to_csv("../data/synthetic_pii_test.csv", index=False)

# Run the engine and print what it found
res = engine.run_rules(syn)
print(res["total_flags"], "flags,", len(res["errors"]), "errors")
for f in res["flags"]:
    print(f["severity"], f["column"], f["rule"], f["confidence"], "|", f["evidence"])
```

```python
# Answer key: did the engine catch what was planted in the synthetic file?
found = {(f["column"], f["rule"]) for f in res["flags"]}
key = pd.DataFrame([
    ("email", "pii", "fake email addresses"),
    ("phone", "pii", "fake phone numbers"),
    ("full_name", "pii", "fake names"),
    ("date_of_birth", "format_inconsistency", "second date format (row 7)"),
    ("county_fips", "format_inconsistency", "lost leading zero (row 15)"),
    ("claim_amount", "extreme_skew", "extreme outlier (row 20)"),
    ("crop", "", "missing value (row 25), not an engine rule"),
], columns=["Column", "Rule expected", "Planted"])
key["Caught?"] = ["Yes" if (c, r) in found else "No" if r else "n/a"
                  for c, r in zip(key["Column"], key["Rule expected"])]
key
```
